```python
import numpy as np
scores = np.array([45,67,72,59,88,34])
scores[0] = 50
print(scores) #[50 67 72 59 88 34]

scores[1:3] = [70,75]
print(scores) #[50 70 75 59 88 34]

scores[scores < 40] = 40 #[50 70 75 59 88 40]
print(scores)

scores[[3,4]] = [60,90] #[50 70 75 60 90 40] 
print(scores)

 
```
