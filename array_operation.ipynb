```python
import numpy as np

array_a = np.array([10,20,30,40])
array_b = np.array([1,2,3,4])

# single array operation
sum = array_a + 5
print(sum)

sub = array_a - 5
print(sub)

mul = array_a * 2
print(mul)

div = array_a /10
print(div)

#two array operation
array_sum = array_a + array_b
print(array_sum)

array_sub = array_a - array_b
print(array_sub)

array_mul = array_a * array_b
print(array_mul)

array_div = array_a / array_b
print(array_div)



```
