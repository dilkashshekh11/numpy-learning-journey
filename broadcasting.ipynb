```python
import numpy as np
test_array = np.array([  [1,2,3],
                         [1,2,3],   
                         [1,2,3]  ])
test_array
```

```python
# scalar : 1 is added to every element
test_array + 1
```

```python
# Row vector (3,): stretched down the rows
print(test_array)
test_array + np.array([3,2,1])
```

```python
# Column vector (3,1): stretched down the columns
print(test_array)
test_array + np.array([ [3],
                        [2],
                        [1] ])
```

```python
# Row vector + Column vector : both stretch to (3,3)
np.array( [1,2,3] ) + np.array([  [2],
                                  [2],
                                  [2]  ])
```

```python
# Q2. Grace marks per SUBJECT: 3 students x 4 subjects
# Add [2, 0, 5, 3] so each subject gets its own grac marks 
marks = np.array([  [45, 60, 38, 70],
                    [55, 48, 62, 40],
                    [30, 75, 50, 66] ])

marks + np.array([2,0,5,3])
```

```python

# Q3. Bonus per STUDENT: add 5, 0 and 10 marks to the 3 students (same marks array)
#
# Hint: the bonus must be a column
marks = np.array([  [45, 60, 38, 70],
                    [55, 48, 62, 40],
                    [30, 75, 50, 66]  ])
marks + np.array([
                [5],
                [0],
                [10]
                    ])
```

```python

# Q4. Compare with average: subtract each subject's average from its marks 
# (positive = above average, negative = below average)
import numpy as np
marks = np.array([   [45, 60, 38, 70],
                     [55, 48, 62, 40],
                     [30, 75, 50, 66]  ])
avg = marks.mean(axis = 0)
print(f"Average = {avg}")

performances =  marks - avg
performances 
```
