```python
# 1D Array
import numpy as np
sales = np.array([120,200,150,180,190,170])
print(f" Sum = {sales.sum()}")


mean_sales = sales.mean()
result = np.round(mean_sales,2)
print(f" Mean = {result}")

print(f" Max = {sales.max()}")

print(f" Min = {sales.min()}")
```

```python
# 2D Array
# 2D aggregation: 3 stores (rows) x 4 quarters (columns)
# axis= -> down the rows, one result per column | axis=1 -> across columns, one result per row
import numpy as np
sales = np.array([  [120, 150, 180, 200],
                    [100, 170, 160, 210],
                    [140, 130, 190, 220]  ])

print(sales)

# whole array sum
print(f"Sum = {sales.sum()}")

# column -wise sum (axis = 0)
print(f"Sum (Column-wise: Axis= 0) = {sales.sum(axis = 0)}")

# rows -wise sum (axis = 1)
print(f"Sum (Rows-wise: Axis = 1) ={sales.sum(axis = 1)} ")

# mean columns and rows-wise
#rows
print(f"Mean columns-wise(axis = 0) = {sales.mean(axis = 0)}")

#rows
print(f"Mean Rows-wise(axis = 1) = {sales.mean(axis = 1)}")



# min columns and rows-wise
#rows
print(f"Min columns-wise(axis = 0) = {sales.min(axis = 0)}")

#rows
print(f"Min Rows-wise(axis = 1) = {sales.min(axis = 1)}")


# min columns and rows-wise
#rows
print(f"Max columns-wise(axis = 0) = {sales.max(axis = 0)}")

#rows
print(f"Max Rows-wise(axis = 1) = {sales.max(axis = 1)}")

```

```python
# 1D Array
import numpy as np
sales = np.array([120,150,180,200,210,220,260])

sales.sort
print(sales)

# percentile
print(f"Percentile 25 = {np.percentile(sales,25)}")
print(f"Percentile 50 = {np.percentile(sales,50)}")
print(f"Percentile 75 = {np.percentile(sales,75)}")

# several percentile at once (25th,50th = median, 75th)
print(f"Percentile (25,50,75) = {np.percentile(sales,[25,50,75])}")
```

```python
import numpy as np
sales = np.array([150,200,150,300,250,200,350,150]) # duplicate values
np.unique(sales)  # array([150, 200, 250, 300, 350])

#unique values And how many times each appears
np.unique(sales, return_counts = True)
```

```python
#square root of each values
np.sqrt(sales)
```

```python
# stardard deviation: std
sales = np.array([120,200,150,180,210,190,170])
sales.std()
```

```python
# postion (index) of the highest and lowest sale
# argmax - for maximum number index
# argmin - for minimum number index
print(sales.argmax(),sales.argmin())
```
