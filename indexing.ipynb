```python
# 1D
import numpy as np
city_array = np.array([
"Mumbai", "Delhi", "Bengaluru", "Hyderabad", "Chennai", "Pune", "Jaipur", "Ahmedabad", "Kolkata", "Lucknow"])
print(f"city_array[0] = {city_array[0]}")
print(f"city_array[-1] = {city_array[-1]}")
# print(f"city_array[-1] = {city_array[100]}") #error
print(f"city_array[5:9] = {city_array[5:9]}")
print(f"city_array[5:] = {city_array[5:]}")
print(f"city_array[:] = {city_array[:]}")
print(f"city_array[1::3] = {city_array[1::3]}")
print(f"city_array[::-1] = {city_array[::-1]}")
```

```python
# 2D
import numpy as np
city_array = np.array(["Mumbai", "Delhi", "Bengaluru", "Hyderabad", "Chennai","Pune", "Jaipur", "Ahmedabad", "Kolkata", "Lucknow"])
dms = city_array.reshape(2,5)
print(dms)
print(f"dms[1,3] = {dms[1,3]}")
print(f"dms[1,-1] = {dms[1,-1]}")
print(f"dms[:,1:4] = {dms[:,1:4]}")
print(f"dms[0,:]= {dms[0,:]}") #for row
print(f"dms[1,:]= {dms[:,4]}") # for column
print(f"dms[1,::-1]= {dms[1:,::-1]}") # fir a row in reverse order
print(f"dms[0,::-1]= {dms[0,::-1]}")  
print()
print(f"dms[:,[3,4]]= {dms[:,[3,4]]}") 
```
