```python
#1D Array
import numpy as np 
temperature = [1,2,3,4,5,6,7,8,9]
temp_array = np.array(temperature)
print(temp_array)
# temp_array
 
print(temp_array.ndim) #type of array
print(temp_array.size) # toal numver of elements
print(temp_array.shape) # shape of array (no.rows,)
print(temp_array.dtype) # dataType of array (like.int64)
```

```python
#2D array
import numpy as np

temperature = [ [1,2,3],
               [4,5,6] ]
temp_array = np.array(temperature)
print(temp_array)

print(temp_array.ndim) #type of array
print(temp_array.size) # toal numver of elements
print(temp_array.shape) # shape of array (no.rows,no.cols)
print(temp_array.dtype) # dataType of array (like.int64)
```
