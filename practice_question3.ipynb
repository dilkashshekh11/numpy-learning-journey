```python

# A teacher has 8 quiz scores and wants to practice NumPy on them. Complete the following tasks.

# Create a 1D NumPy array named "scores" using:
# 16, 25, 36, 25, 49, 64, 81, 100.
import numpy as np
scores = np.array([16,25,36,49,64,81,100])


# Update every score below 25 to 25.
result = np.where(scores < 25 , 25,scores)
print(result)

# Create a new array named "sorted_scores" by sorting the updated scores in ascending order.

sorted_scores = np.sort(scores)
print(sorted_scores)

# From the updated array, find the total, average, highest value, and median.
print(f"Total = {np.sum(sorted_scores)}")
print(f"Average = {np.mean(sorted_scores)}")
print(f"Highest Value = {np.max(sorted_scores)}")
print(f"Median = {np.median(sorted_scores)}")

# Use "np.unique()" to get the distinct score values.
print(f"Unique Value = {np.unique(sorted_scores)}")

# Use "np.sqrt()" to calculate the square root of each updated score.

print(f"Square root = {np.sqrt(sorted_scores)}")
```
