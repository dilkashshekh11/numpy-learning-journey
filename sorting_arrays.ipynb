```python
# 1D array
import numpy as np

marks = np.array([72,45,88,61,95,53])
res = np.sort(marks)
print(f"Sorted = {res}")
# print(marks)
```

```python
# 2D array
sales_2d = np.array([
[52, 18, 41, 77, 29, 63], 
[34, 91, 25, 48, 56, 12], 
[83, 39, 67, 21, 45, 30], 
[14, 72, 58, 36, 90, 27]
])
#rows-wise sorted
sales_2d.sort(axis = 1)
print(f"Rows-wise sorted(axis = 1) =\n{sales_2d}")
```

```python
# 2D array
sales_2d = np.array([
[52, 18, 41, 77, 29, 63], 
[34, 91, 25, 48, 56, 12], 
[83, 39, 67, 21, 45, 30], 
[14, 72, 58, 36, 90, 27]
])
#columns-wise sorted
sales_2d.sort(axis = 0)
print(f"Columns-wise sorted (axis = 0) = \n{sales_2d}")
```
