```python
#
#Q6. (Challenge) Stock report using yesterday's and today's prices: First find the % change (today- yesterday) / yesterday * 100

# Then label: "Big Jump" if change > 5, "Up" if change > 0,
# "Down" if change < 0, else "No Change"
# Hint: put one np. where inside another

yesterday= np.array([200, 150, 320, 90, 500, 75]) 
today= np.array([215, 150, 310, 93, 500, 72])
changes = (today - yesterday) / yesterday * 100

print(changes)

stock_report = np.where(changes > 0 , "Big Jump",
                        np.where(changes > 0,"Up", 
                                 np.where(changes < 0,"Down","No Change")))
print(stock_report)
```

```python

# Practise Question: A teacher has exam scores for 8 students and wants to analyze them with NumPy. Complete the following tasks:
import numpy as np
marks = np.array([35, 42, 50, 61, 28, 74, 81, 39])
print(marks)
# Create a 2 x 4 NumPy array called scores using these values:
scores = marks.reshape(2,4)
print(f" 2 x 4 Array  = {scores}")

# 35, 42, 50, 61, 28, 74, 81, 39.
# Add 5 grace marks to every score and store the result in updated scores.
updated_scores = scores + 5
print(f" Adding + 5 ={updated_scores}")

# From updated scores, extract only the second row.
print(f"'Extract 2nd Row'/updated_scores[1:] = {updated_scores[1:]}")
# Filter the values in updated scores that are 50 or higher.

higher_scores = updated_scores[updated_scores >= 50]
print(f"Higher Score  = {higher_scores}")

# Use np. where() to create a new array that shows 'Pass' for scores 50 or more and 'Needs Work' for scores below 50.
final_array = np.where(updated_scores >= 50, "Pass","Needs Work")
print(f"Final Array = {final_array}")
```
