```python
# problme given: in normal python form
# We wantt to multipy two lists elements-wise, may times (10,000 loops).
# Create two lists.


list1= list(range(10000))
list2 = list(range(10000))

# Multiply using a for loop (element by element) 
def for_loop_multiply_lists(l1, l2):
    product_list = []
    for i in range(len(l1)):
        product_list.append(l1[i] * l2[i])
    return product_list

result = for_loop_multiply_lists(list1,list2)
print(result[:10])

```

```python
# Solution: using Numpy arrays and vactorized operations
import numpy as np
list1= list(range(10000))
list2 = list(range(10000))

first_array = np.array(list1)
second_array = np.array(list2)

# multipy both array : element-wise 

result = first_array * second_array
print(f"Multiplication of both arrays: {result}")
# result
```
